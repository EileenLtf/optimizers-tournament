```python
import numpy as np 
from numpy.random import Generator as gen 
from numpy.random import PCG64 as pcg
import matplotlib.pyplot as plt 

seed = 101 
```

```python
def Generate_dots(a): # tool/arz square
    array = gen(pcg(seed))
    return array.uniform(-a, a, size=(2000, 2))
     
```

```python
array1 = Generate_dots(1.5)
array1 
```

```python
def labeling_circle_boundry(r, data):
    labels = (data[:, 0]**2 + data[:, 1] **2 < r**2).astype(int)
    return labels 
```

```python
labels = labeling_circle_boundry(1, array1)
```

```python
labels 
```

```python
fig, ax = plt.subplots(figsize=(7, 7))
ax.scatter(array1[:, 0], array1[:, 1], c=labels )
circle = plt.Circle((0, 0), 1, fill=False, color="red",linewidth=2.5) 
ax.add_patch(circle)
ax.set_title("in cirlcle or out of it?")
ax.set_xlabels("x1")
ax.set_ylabels("x2")
ax.set_aspect("equal")
ax.grid(True, alpha=0.3)

plt.show()

return fig, ax
```
